```python
# ============================================================
# ML Project:
# Regression + Classification + Preprocessing + Imbalance Handling
# Dataset: Ames Housing Dataset
# Fixed Version: Does NOT rely only on OpenML
# ============================================================

!pip install -q imbalanced-learn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import TransformedTargetRegressor

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from sklearn.linear_model import LinearRegression, Ridge, LogisticRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier,
    GradientBoostingRegressor,
    GradientBoostingClassifier,
    ExtraTreesRegressor,
    ExtraTreesClassifier
)
from sklearn.svm import SVR, SVC
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from imblearn.over_sampling import RandomOverSampler


# ============================================================
# 1. Load Dataset Safely
# ============================================================

def load_ames_housing():
    urls = [
        "https://raw.githubusercontent.com/jcp9010/MSDA/master/CUNY%20605/Final%20Project/train.csv",
        "https://raw.githubusercontent.com/eddiexunyc/DATA605_FINAL/main/Resources/train.csv",
        "https://raw.githubusercontent.com/chriskhanhtran/kaggle-house-price/master/train.csv"
    ]

    last_error = None

    for url in urls:
        try:
            print("Trying to load dataset from:")
            print(url)

            df = pd.read_csv(url)

            if "SalePrice" in df.columns:
                print("\nDataset loaded successfully.")
                return df

        except Exception as e:
            last_error = e
            print("This link failed. Trying another one...\n")

    raise Exception(f"All dataset links failed. Last error: {last_error}")


df = load_ames_housing()

print("\nDataset Shape:", df.shape)
display(df.head())

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
missing_values = df.isnull().sum()
display(missing_values[missing_values > 0].sort_values(ascending=False))


# ============================================================
# 2. Basic Cleaning
# ============================================================

if "Id" in df.columns:
    df = df.drop(columns=["Id"])

target_col = "SalePrice"

df[target_col] = pd.to_numeric(df[target_col], errors="coerce")
df = df.dropna(subset=[target_col])

X = df.drop(columns=[target_col])
y_reg = df[target_col]

print("\nRegression Target Summary:")
display(y_reg.describe())


# ============================================================
# 3. Detect Numeric and Categorical Features
# ============================================================

numeric_features = X.select_dtypes(include=["int64", "float64", "int32", "float32"]).columns.tolist()
categorical_features = X.select_dtypes(exclude=["int64", "float64", "int32", "float32"]).columns.tolist()

print("\nNumber of Numeric Features:", len(numeric_features))
print("Number of Categorical Features:", len(categorical_features))


# ============================================================
# 4. Preprocessing Pipeline
# ============================================================

numeric_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler())
])

try:
    categorical_encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:
    categorical_encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)

categorical_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", categorical_encoder)
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_preprocessor, numeric_features),
        ("cat", categorical_preprocessor, categorical_features)
    ],
    sparse_threshold=0.0
)


# ============================================================
# 5. Regression Train-Test Split
# ============================================================

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_reg,
    test_size=0.2,
    random_state=42
)


# ============================================================
# 6. Regression Models
# ============================================================

regression_models = {
    "Linear Regression": LinearRegression(),

    "Ridge Regression": Ridge(alpha=10),

    "KNN Regressor": KNeighborsRegressor(
        n_neighbors=7
    ),

    "SVR": SVR(
        kernel="rbf",
        C=100,
        epsilon=0.1,
        gamma="scale"
    ),

    "Random Forest Regressor": RandomForestRegressor(
        n_estimators=250,
        max_depth=None,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees Regressor": ExtraTreesRegressor(
        n_estimators=250,
        max_depth=None,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting Regressor": GradientBoostingRegressor(
        n_estimators=250,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}


# ============================================================
# 7. Train and Evaluate Regression Models
# ============================================================

regression_results = []

best_reg_model_name = None
best_reg_rmse = float("inf")
best_reg_pipeline = None
best_reg_predictions = None

for name, model in regression_models.items():
    print(f"\nTraining Regression Model: {name}")

    base_pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    reg_pipeline = TransformedTargetRegressor(
        regressor=base_pipeline,
        func=np.log1p,
        inverse_func=np.expm1
    )

    reg_pipeline.fit(X_train_reg, y_train_reg)
    y_pred = reg_pipeline.predict(X_test_reg)

    mse = mean_squared_error(y_test_reg, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test_reg, y_pred)
    r2 = r2_score(y_test_reg, y_pred)
    mape = np.mean(np.abs((y_test_reg - y_pred) / y_test_reg)) * 100

    regression_results.append({
        "Model": name,
        "MSE": mse,
        "RMSE": rmse,
        "MAE": mae,
        "R2 Score": r2,
        "MAPE %": mape
    })

    if rmse < best_reg_rmse:
        best_reg_rmse = rmse
        best_reg_model_name = name
        best_reg_pipeline = reg_pipeline
        best_reg_predictions = y_pred

regression_results_df = pd.DataFrame(regression_results)
regression_results_df = regression_results_df.sort_values(by="RMSE")

print("\n================ REGRESSION RESULTS ================")
display(regression_results_df)

print("\nBest Regression Model:", best_reg_model_name)


# ============================================================
# 8. Regression Plots
# ============================================================

plt.figure(figsize=(8, 6))
plt.scatter(y_test_reg, best_reg_predictions, alpha=0.5)
plt.xlabel("Actual Sale Price")
plt.ylabel("Predicted Sale Price")
plt.title(f"Actual vs Predicted Sale Price - {best_reg_model_name}")
plt.grid(True)
plt.show()

plt.figure(figsize=(9, 5))
plt.barh(regression_results_df["Model"], regression_results_df["RMSE"])
plt.xlabel("RMSE")
plt.title("Regression Model Comparison - RMSE")
plt.gca().invert_yaxis()
plt.grid(axis="x")
plt.show()

residuals = y_test_reg - best_reg_predictions

plt.figure(figsize=(8, 5))
plt.scatter(best_reg_predictions, residuals, alpha=0.5)
plt.axhline(y=0, linestyle="--")
plt.xlabel("Predicted Sale Price")
plt.ylabel("Residuals")
plt.title(f"Residual Plot - {best_reg_model_name}")
plt.grid(True)
plt.show()


# ============================================================
# 9. Convert Regression Target into Classification Target
# ============================================================

def price_to_class(price):
    if price < 150000:
        return 0
    elif price < 250000:
        return 1
    else:
        return 2

y_class = y_reg.apply(price_to_class)

class_names = [
    "Affordable House",
    "Standard House",
    "Expensive House"
]

print("\nClassification Target Distribution:")

class_counts = y_class.value_counts().sort_index()

for class_id, count in class_counts.items():
    print(f"{class_id} - {class_names[class_id]}: {count}")

plt.figure(figsize=(7, 5))
plt.bar([class_names[i] for i in class_counts.index], class_counts.values)
plt.xlabel("House Price Class")
plt.ylabel("Number of Houses")
plt.title("Class Distribution")
plt.grid(axis="y")
plt.show()


# ============================================================
# 10. Check for Real Imbalance
# ============================================================

max_class_count = class_counts.max()
min_class_count = class_counts.min()
imbalance_ratio = max_class_count / min_class_count

print("\nImbalance Ratio:", round(imbalance_ratio, 3))

if imbalance_ratio >= 1.5:
    imbalance_exists = True
    print("Imbalance detected. Imbalance handling will be applied.")
else:
    imbalance_exists = False
    print("No serious imbalance detected. Normal training will be used.")


# ============================================================
# 11. Classification Train-Test Split
# ============================================================

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X,
    y_class,
    test_size=0.2,
    random_state=42,
    stratify=y_class
)


# ============================================================
# 12. Classification Models
# ============================================================

if imbalance_exists:
    classification_models = {
        "Logistic Regression with Class Weight": LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=42
        ),

        "SVC with Class Weight": SVC(
            kernel="rbf",
            C=10,
            gamma="scale",
            class_weight="balanced",
            random_state=42
        ),

        "Random Forest with Class Weight": RandomForestClassifier(
            n_estimators=250,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

        "Extra Trees with Class Weight": ExtraTreesClassifier(
            n_estimators=250,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),

        "Gradient Boosting with Oversampling": GradientBoostingClassifier(
            n_estimators=250,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

        "KNN with Oversampling": KNeighborsClassifier(
            n_neighbors=7
        )
    }

else:
    classification_models = {
        "Logistic Regression": LogisticRegression(
            max_iter=3000,
            random_state=42
        ),

        "SVC": SVC(
            kernel="rbf",
            C=10,
            gamma="scale",
            random_state=42
        ),

        "Random Forest Classifier": RandomForestClassifier(
            n_estimators=250,
            max_depth=None,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ),

        "Extra Trees Classifier": ExtraTreesClassifier(
            n_estimators=250,
            max_depth=None,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        ),

        "Gradient Boosting Classifier": GradientBoostingClassifier(
            n_estimators=250,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

        "KNN Classifier": KNeighborsClassifier(
            n_neighbors=7
        )
    }


# ============================================================
# 13. Train and Evaluate Classification Models
# ============================================================

classification_results = []

best_cls_model_name = None
best_cls_f1 = -1
best_cls_pipeline = None
best_cls_predictions = None

for name, model in classification_models.items():
    print(f"\nTraining Classification Model: {name}")

    if imbalance_exists and "Oversampling" in name:
        cls_pipeline = ImbPipeline(steps=[
            ("preprocessor", preprocessor),
            ("oversampler", RandomOverSampler(random_state=42)),
            ("model", model)
        ])
    else:
        cls_pipeline = Pipeline(steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ])

    cls_pipeline.fit(X_train_cls, y_train_cls)
    y_pred = cls_pipeline.predict(X_test_cls)

    accuracy = accuracy_score(y_test_cls, y_pred)

    precision_macro = precision_score(
        y_test_cls,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall_macro = recall_score(
        y_test_cls,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1_macro = f1_score(
        y_test_cls,
        y_pred,
        average="macro",
        zero_division=0
    )

    precision_weighted = precision_score(
        y_test_cls,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall_weighted = recall_score(
        y_test_cls,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1_weighted = f1_score(
        y_test_cls,
        y_pred,
        average="weighted",
        zero_division=0
    )

    classification_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision Macro": precision_macro,
        "Recall Macro": recall_macro,
        "F1 Macro": f1_macro,
        "Precision Weighted": precision_weighted,
        "Recall Weighted": recall_weighted,
        "F1 Weighted": f1_weighted
    })

    if f1_macro > best_cls_f1:
        best_cls_f1 = f1_macro
        best_cls_model_name = name
        best_cls_pipeline = cls_pipeline
        best_cls_predictions = y_pred

classification_results_df = pd.DataFrame(classification_results)
classification_results_df = classification_results_df.sort_values(
    by="F1 Macro",
    ascending=False
)

print("\n================ CLASSIFICATION RESULTS ================")
display(classification_results_df)

print("\nBest Classification Model:", best_cls_model_name)


# ============================================================
# 14. Classification Report
# ============================================================

print("\nClassification Report for Best Model:")
print(classification_report(
    y_test_cls,
    best_cls_predictions,
    target_names=class_names,
    zero_division=0
))


# ============================================================
# 15. Confusion Matrix
# ============================================================

cm = confusion_matrix(y_test_cls, best_cls_predictions)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

fig, ax = plt.subplots(figsize=(8, 6))
disp.plot(ax=ax)
plt.title(f"Confusion Matrix - {best_cls_model_name}")
plt.xticks(rotation=20)
plt.show()


# ============================================================
# 16. Classification Model Comparison Plot
# ============================================================

plt.figure(figsize=(10, 5))
plt.barh(classification_results_df["Model"], classification_results_df["F1 Macro"])
plt.xlabel("Macro F1 Score")
plt.title("Classification Model Comparison - Macro F1")
plt.gca().invert_yaxis()
plt.grid(axis="x")
plt.show()


# ============================================================
# 17. Feature Importance
# ============================================================

def get_feature_names_from_preprocessor(fitted_preprocessor):
    feature_names = []

    if len(numeric_features) > 0:
        feature_names.extend(numeric_features)

    if len(categorical_features) > 0:
        encoder = fitted_preprocessor.named_transformers_["cat"].named_steps["encoder"]
        encoded_features = encoder.get_feature_names_out(categorical_features)
        feature_names.extend(encoded_features)

    return feature_names


def plot_feature_importance_from_pipeline(pipeline, model_name):
    try:
        fitted_preprocessor = pipeline.named_steps["preprocessor"]
        model = pipeline.named_steps["model"]

        if hasattr(model, "feature_importances_"):
            feature_names = get_feature_names_from_preprocessor(fitted_preprocessor)

            importance_df = pd.DataFrame({
                "Feature": feature_names,
                "Importance": model.feature_importances_
            })

            importance_df = importance_df.sort_values(
                by="Importance",
                ascending=False
            ).head(20)

            print(f"\nTop 20 Feature Importances - {model_name}")
            display(importance_df)

            plt.figure(figsize=(10, 6))
            plt.barh(importance_df["Feature"], importance_df["Importance"])
            plt.xlabel("Importance")
            plt.title(f"Top 20 Feature Importances - {model_name}")
            plt.gca().invert_yaxis()
            plt.grid(axis="x")
            plt.show()

        else:
            print(f"\n{model_name} does not provide feature_importances_.")

    except Exception as e:
        print("Could not plot feature importance:", e)


plot_feature_importance_from_pipeline(best_cls_pipeline, best_cls_model_name)


# ============================================================
# 18. Final Project Summary
# ============================================================

print("\n================ FINAL PROJECT SUMMARY ================")

print("\nDataset:")
print("Ames Housing Dataset")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nRegression Task:")
print("Target: SalePrice")
print("Best Regression Model:", best_reg_model_name)
print("Best Regression RMSE:", round(best_reg_rmse, 3))

print("\nClassification Task:")
print("Target: House Price Class")
print("Classes:")
print("0 = Affordable House, SalePrice < 150,000")
print("1 = Standard House, 150,000 <= SalePrice < 250,000")
print("2 = Expensive House, SalePrice >= 250,000")

print("\nBest Classification Model:", best_cls_model_name)
print("Best Classification Macro F1:", round(best_cls_f1, 4))

print("\nImbalance Handling:")
if imbalance_exists:
    print("Class imbalance was detected naturally after converting SalePrice into classes.")
    print("The code handled it using class weights and oversampling where suitable.")
else:
    print("No serious class imbalance was detected.")
    print("The models were trained normally.")

print("\nPreprocessing Used:")
print("- Dropped Id column")
print("- Missing numerical values handled with median imputation")
print("- Missing categorical values handled with most frequent imputation")
print("- Numerical features scaled using RobustScaler")
print("- Categorical features encoded using OneHotEncoder")
print("- Regression target transformed using log1p for better stability")

print("\nEvaluation Metrics:")
print("- Regression: MSE, RMSE, MAE, R2 Score, MAPE")
print("- Classification: Accuracy, Precision, Recall, F1 Score, Confusion Matrix")
```
